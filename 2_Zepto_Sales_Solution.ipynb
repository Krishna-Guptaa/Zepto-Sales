### **Influencer Count (Active or Not)**

```python
a = ["royalblue","darkslateblue"]

sns.countplot(x = "Influencer Active", data =df, palette = a)

plt.title("Count of Active Influencers", color = "red")
plt.xlabel("Active or Not", color = "navy")
plt.ylabel("Count of Influencers", color = "navy")

plt.legend()
plt.show()
```

### **Visualisation of top 3 Product by Total Revenue**

```python
total_revenue = df.nlargest(3, ["Total Revenue"])

bar_colors = ["deeppink", "teal", "mediumseagreen"]

sns.barplot(x = total_revenue["Product Name"], y = total_revenue["Total Revenue"], palette = bar_colors)

plt.title("Top 3 Products by Total Revenue", color = "navy")
plt.xlabel("Name of the Products", color = "darkslategrey")
plt.ylabel("Sum of Revenue", color = "darkslategrey")
plt.show()
```

### **Sum of Current Price by Cities**

```python
group_city = df.groupby("City")["Current Price"].sum()

sns.barplot(x = group_city.index, y = group_city.values, palette = "crest")

plt.title("Sum of Product Price by City", color = "purple")
plt.xlabel("City Name", color = "dimgrey")
plt.ylabel("Sum of Price", color = "dimgrey")
plt.show()
```

### **Order by Original Price with Active Influencers**

```python
sns.scatterplot(x = "Original Price", y = "Orders", data = df, hue = "Influencer Active")
```

### **Product Placed with Current price**

```python
plt.figure(figsize = (8,5))

bubble_color = ["rosybrown", "brown"]

sns.stripplot(x ="Category", y = "Current Price", data = df, hue ="Influencer Active", palette = bubble_color )

plt.title("How many Product were placed respective their Current Price")
plt.xlabel("Categories")
plt.ylabel("Current Price")

plt.show()
```

### **Discount by Product**

```python
pie_chart = df.groupby("Product Name")["Discount"].sum()

plt.pie(pie_chart, labels = pie_chart.index, autopct = "%1.1f%%", colors = [ "gray", "lightblue", "pink", "orange", "green","purple", "yellow", "cyan", "red", "brown"])

plt.title("Discount by Product", color = "red")
plt.legend(fontsize = 7)
plt.show()
```

### **Total Revenue by Category**

```python
plt.Figure(figsize = (8,5))

sns.lineplot(x = "Category", y = "Total Revenue", data = df )

plt.title("Total Revenue by Category", color = "crimson")
```

```python
df.groupby("Category")["Current Price"].mean().sort_values()
```

```python
avg_price = df.groupby("Category")["Current Price"].mean().sort_values()

plt.figure(figsize=(10, 6))

plt.bar(avg_price.index, avg_price.values, color = ["lightcoral","indianred", "brown", "firebrick","maroon","darkred"])

plt.title("Average Price by Category", color = "teal")
plt.xlabel("Category", color = "orchid")
plt.ylabel("Average Price", color = "orchid")

plt.show()
```
